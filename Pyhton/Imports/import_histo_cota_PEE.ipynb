# Import historique des livret PEE STEF (CSV) et CiC (Excel)

```python
import pandas as pd
from sqlalchemy import create_engine
import config
from datetime import datetime

# 1. Connection à la DB
url = f"postgresql://{config.db_params['user']}:{config.db_params['password']}@{config.db_params['host']}:{config.db_params['port']}/{config.db_params['database']}"
engine = create_engine(url)

print("\nConnexion établie !")
```

### STEF

```python
file = r"E:\Data\Tristan\Téléchargements\HistoriqueValeurPart_26_01_2021_26_01_2026.csv"

df_stef = pd.read_csv(
    file, 
    sep=';', 
    header=0,                                     # On ignore la ligne de titres d'origine
    names=['cot_date_prix', 'cot_prix_unitaire'], # On impose NOS noms de colonnes SQL
    usecols=[0, 1],
    parse_dates=['cot_date_prix'], 
    dayfirst=True)

df_stef['pdt_id'] = 3

df_stef
```

```python
df_stef.to_sql('cotation_cot', engine, if_exists='append', index=False)
print("STEF importé")
```

### CiC

#### Livret Stratégie

```python
start = '2022-01-01'

file = r"E:\Data\Tristan\Téléchargements\Historique Cours Strategie.xlsx"

df_cic_strat = pd.read_excel(
    file, 
    skiprows=6,                                   # On exclue des lignes au debut du fichier
    skipfooter=1,                                 # On exclue des lignes à la fin du fichier
    names=['cot_date_prix', 'cot_prix_unitaire'], # On impose NOS noms de colonnes SQL
    usecols=[0, 1],
    parse_dates=['cot_date_prix'],
    date_format="%d/%m/%Y")

df_cic_strat['pdt_id'] = 6
df_cic_strat = df_cic_strat[df_cic_strat['cot_date_prix'] >= start]

df_cic_strat
```

```python
df_cic_strat.to_sql('cotation_cot', engine, if_exists='append', index=False)
print("CiC Strat importé")
```
